### 1. Basic Arithmetic Operations (Addition, Subtraction, Multiplication, Division)

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

sum = a + b
difference = a - b
product = a * b
division = a / b

print("Sum = ", sum)
print("Difference = ", difference)
print("Product = ", product)
print("Division = ", division)
```

### 2. Check Even or Odd Number

```python
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is Even")
else:
    print(num, "is Odd")

print("Program Finished")
```

### 3. Find the Largest Among Three Numbers

```python
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    print("Largest =", a)
elif b >= c:
    print("Largest =", b)
else:
    print("Largest =", c)
```

### 4. Multiplication Table Generator

```python
num = int(input("Enter a number: "))

print("Multiplication Table")

for i in range(1, 11):
    result = num * i
    print(num, "x", i, "=", result)

print("Completed")
```

### 5. Find Factorial of a Number (Using a while Loop)

```python
num = int(input("Enter a number: "))

fact = 1
i = 1

while i <= num:
    fact = fact * i
    i += 1

print("Factorial =", fact)
```

### 6. Check Prime Number

```python
num = int(input("Enter a number: "))

count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count += 1

if count == 2:
    print("Prime Number")
else:
    print("Not Prime")
```

### 7. Basic List Operations (Append, Remove, Length, and Indexing)

```python
numbers = [10, 20, 30, 40, 50]

numbers.append(60)
numbers.remove(20)

print("Updated List:", numbers)
print("Length:", len(numbers))
print("First Element:", numbers[0])
print("Last Element:", numbers[-1])
```

### 8. Dictionary Manipulation

```python
student = {
    "Name": "Rahul",
    "Age": 20,
    "Branch": "CSE"
}

print("Name:", student["Name"])
print("Age:", student["Age"])

student["Age"] = 21

print("Updated Dictionary:", student)
```

### 9. Square Calculation Using Functions

```python
def square(number):
    result = number * number
    return result

num = int(input("Enter a number: "))

answer = square(num)

print("Square =", answer)
```
