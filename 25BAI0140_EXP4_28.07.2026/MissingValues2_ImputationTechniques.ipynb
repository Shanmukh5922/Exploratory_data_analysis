```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer

df_new = pd.DataFrame(
    {
        "Feature1": [10, np.nan, 15, 20, 12],
        "Feature2": [100, 120, np.nan, 150, 110],
        "Feature3": ["X", "Y", "X", np.nan, "Y"],
    }
)

mean_imputer_new = SimpleImputer(strategy="mean")
df_new["Feature1_Mean"] = mean_imputer_new.fit_transform(df_new[["Feature1"]])

median_imputer_new = SimpleImputer(strategy="median")
df_new["Feature2_Median"] = median_imputer_new.fit_transform(df_new[["Feature2"]])

mode_imputer_new = SimpleImputer(strategy="most_frequent")
df_new["Feature3_Mode"] = mode_imputer_new.fit_transform(df_new[["Feature3"]])[:, 0]

num_cols_new = ["Feature1", "Feature2"]
mice_imputer_new = IterativeImputer(max_iter=10, random_state=42)
df_mice_new = df_new[num_cols_new].copy()
df_mice_new.iloc[:, :] = mice_imputer_new.fit_transform(df_mice_new)
df_new["Feature1_MICE"] = df_mice_new["Feature1"]
df_new["Feature2_MICE"] = df_mice_new["Feature2"]
print(df_new)
```

## Missing Values

```python
data_new = {
    'Product': ['A', 'B', 'C', 'D', 'E', 'F'],
    'Sales_Q1': [100, 150, np.nan, 120, 200, 80],
    'Sales_Q2': [np.nan, 160, 110, 130, np.nan, 90],
    'Region': ['East', 'West', 'North', np.nan, 'East', 'South'],
    'Units_Sold': [10, 15, 11, np.nan, np.nan, 8]
}
df_products = pd.DataFrame(data_new)
df_products.loc[6] = ['G', np.nan, np.nan, np.nan, np.nan]
df_products.loc[7] = [np.nan, np.nan, np.nan, np.nan, np.nan]
df_products['Notes'] = np.nan
display(df_products)
```

```python
display(df_products.isnull())

display(df_products.notnull())
```

```python
print("Number of NaN values per column:")
display(df_products.isnull().sum())

print("Total number of NaN values in the DataFrame:")
display(df_products.isnull().sum().sum())
```

```python
print("Number of reported (non-NaN) values per column:")
display(df_products.count())
```

```python
print("DataFrame after dropping rows with any NaN:")
display(df_products.dropna())

print("DataFrame after dropping rows where ALL values are NaN (how='all'):")
display(df_products.dropna(how='all'))

print("DataFrame after dropping columns where ALL values are NaN (axis=1, how='all'):")
display(df_products.dropna(axis=1, how='all'))

print("DataFrame after dropping rows with less than 4 non-NaN values (thresh=4):")
display(df_products.dropna(thresh=4))
```

```python
print(f"Mean of Sales_Q1: {df_products['Sales_Q1'].mean()}")
print(f"Sum of Sales_Q2: {df_products['Sales_Q2'].sum()}")

print("Cumulative sum for Units_Sold:")
display(df_products['Units_Sold'].cumsum())
```

```python
print("DataFrame after filling NaNs with 0:")
filled_df_zero = df_products.fillna(0)
display(filled_df_zero)

print(f"Mean of Sales_Q1 (original): {df_products['Sales_Q1'].mean()}")
print(f"Mean of Sales_Q1 (filled with 0): {filled_df_zero['Sales_Q1'].mean()}")

print("DataFrame after forward-filling 'Sales_Q2':")
display(df_products['Sales_Q2'].fillna(method='ffill'))

print("DataFrame after backward-filling 'Region':")
display(df_products['Region'].fillna(method='bfill'))
```
