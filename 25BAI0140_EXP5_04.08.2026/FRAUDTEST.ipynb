## Applying Operations on `fraudTest.csv`

```python
import pandas as pd

df_fraud = pd.read_csv('/content/fraudTest.csv')

print("First 5 rows of fraudTest.csv:")
display(df_fraud.head())
```

### Discretization and Binning on 'amt' column

```python
print("Descriptive statistics for 'amt' column:")
display(df_fraud['amt'].describe())

bins_amt = [0, 50, 100, 200, 500, df_fraud['amt'].max() + 1]
labels_amt = ['Very Low', 'Low', 'Medium', 'High', 'Very High']

df_fraud['amt_binned'] = pd.cut(df_fraud['amt'], bins=bins_amt, labels=labels_amt, right=False)

print("\nValue counts for 'amt_binned':")
display(df_fraud['amt_binned'].value_counts())

df_fraud['amt_quartiles'] = pd.qcut(df_fraud['amt'], 4)

print("\nValue counts for 'amt_quartiles' (equal-sized bins):")
display(df_fraud['amt_quartiles'].value_counts())
```

### Outlier Detection and Filtering on 'amt' column

```python
import numpy as np

threshold_amt_outlier = df_fraud['amt'].quantile(0.99)
print(f"Outlier threshold for 'amt' (99th percentile): {threshold_amt_outlier:.2f}")

outliers_amt = df_fraud[df_fraud['amt'] > threshold_amt_outlier]

print(f"\nNumber of outliers detected in 'amt' column (above {threshold_amt_outlier:.2f}): {len(outliers_amt)}")
print("\nFirst 5 rows of transactions identified as outliers:")
display(outliers_amt.head())
```

### Feature Scaling and Normalization

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

numerical_cols = ['amt', 'city_pop', 'unix_time', 'lat', 'long']
df_numerical = df_fraud[numerical_cols].copy()

print("Original Numerical Data (first 5 rows):")
display(df_numerical.head())

min_max_scaler = MinMaxScaler()
df_min_max_scaled = pd.DataFrame(
    min_max_scaler.fit_transform(df_numerical),
    columns=numerical_cols
)
print("\nMin-Max Scaled Data (first 5 rows):")
display(df_min_max_scaled.head())

standard_scaler = StandardScaler()
df_standard_scaled = pd.DataFrame(
    standard_scaler.fit_transform(df_numerical),
    columns=numerical_cols
)
print("\nZ-score Standardized Data (first 5 rows):")
display(df_standard_scaled.head())

robust_scaler = RobustScaler()
df_robust_scaled = pd.DataFrame(
    robust_scaler.fit_transform(df_numerical),
    columns=numerical_cols
)
print("\nRobust Scaled Data (first 5 rows):")
display(df_robust_scaled.head())
```
