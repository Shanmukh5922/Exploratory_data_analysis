## Outlier Detection and Filtering



```python
import numpy as np

if 'qty' in df_fraud.columns:
    df_fraud['TotalPrice'] = df_fraud['amt'] * df_fraud['qty']
    print("Calculated 'TotalPrice' by multiplying 'amt' and 'qty'.")
else:
    df_fraud['TotalPrice'] = df_fraud['amt']
    print("Using 'amt' as 'TotalPrice' for outlier detection.")

print("\n'TotalPrice' column statistics:")
print(df_fraud['TotalPrice'].describe())

Q1 = df_fraud['TotalPrice'].quantile(0.25)
Q3 = df_fraud['TotalPrice'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"\nIQR Outlier Bounds: Lower = {lower_bound:.2f}, Upper = {upper_bound:.2f}")

outliers = df_fraud[(df_fraud['TotalPrice'] < lower_bound) | (df_fraud['TotalPrice'] > upper_bound)]
print(f"\nNumber of outliers detected using IQR: {len(outliers)}")

if not outliers.empty:
    print("\nFirst 5 detected outliers in 'TotalPrice':")
    display(outliers[['amt', 'TotalPrice']].head())

    df_fraud_filtered = df_fraud[(df_fraud['TotalPrice'] >= lower_bound) & (df_fraud['TotalPrice'] <= upper_bound)]
    print(f"\nShape of DataFrame after filtering outliers: {df_fraud_filtered.shape}")
    df_fraud_processed = df_fraud_filtered.copy() # Create a processed DataFrame for subsequent steps
else:
    print("No outliers detected using the IQR method.")
    df_fraud_processed = df_fraud.copy() # If no outliers, continue with the original DataFrame
```

## Feature Scaling and Normalization



```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

amt_data = df_fraud_processed['amt'].dropna().values.reshape(-1, 1)

if amt_data.size > 0:
    min_max_scaler = MinMaxScaler()
    df_fraud_processed['amt_minmax_scaled'] = min_max_scaler.fit_transform(amt_data)
    print("\n'amt' column after Min-Max Scaling (on processed data):")
    display(df_fraud_processed[['amt', 'amt_minmax_scaled']].head())
    print("Min-Max Scaled Data Statistics:")
    print(df_fraud_processed['amt_minmax_scaled'].describe())

    standard_scaler = StandardScaler()
    df_fraud_processed['amt_standard_scaled'] = standard_scaler.fit_transform(amt_data)
    print("\n'amt' column after Z-score Standardization (on processed data):")
    display(df_fraud_processed[['amt', 'amt_standard_scaled']].head())
    print("Standardized Data Statistics:")
    print(df_fraud_processed['amt_standard_scaled'].describe())
else:
    print("'amt' column is empty or contains only NaNs after dropping them. Cannot perform scaling.")
```
