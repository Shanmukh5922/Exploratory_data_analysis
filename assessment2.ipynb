## Merging DataFrames with a New Dataset (Raining Data)

Let's create two new DataFrames, `df_temp` (temperature data) and `df_humidity` (humidity data), to demonstrate the different merging options. Both will have a 'Date' column, and we'll ensure some dates overlap and some don't, to clearly show the effects of different join types.

```python
import pandas as pd


# Temperature data
df_temp = pd.DataFrame({
    'Date': ['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-05', '2023-01-06'],
    'Avg_Temperature_C': [10, 12, 11, 15, 13],
    'City': ['London', 'London', 'London', 'London', 'London']
})

# Humidity data
df_humidity = pd.DataFrame({
    'Date': ['2023-01-02', '2023-01-03', '2023-01-04', '2023-01-06', '2023-01-07'],
    'Avg_Humidity_Perc': [85, 90, 88, 92, 80],
    'City': ['London', 'London', 'London', 'London', 'London']
})

print("df_temp:")
display(df_temp)
print("\ndf_humidity:")
display(df_humidity)
```

### Option 1: Concatenating DataFrames (axis=1)



```python
df_concat_axis1 = pd.concat([df_temp, df_humidity], axis=1)
print("Concatenated DataFrame (axis=1):")
display(df_concat_axis1)
```

### Option 2: Inner Merge



```python
df_inner_merge = df_temp.merge(df_humidity, on='Date', how='inner')
print("Inner Merged DataFrame:")
display(df_inner_merge)
```

### Option 3: Left Merge



```python
df_left_merge = df_temp.merge(df_humidity, on='Date', how='left')
print("Left Merged DataFrame:")
display(df_left_merge)
```

### Option 4: Right Merge



```python
df_right_merge = df_temp.merge(df_humidity, on='Date', how='right')
print("Right Merged DataFrame:")
display(df_right_merge)
```

## Reshaping with Hierarchical Indexing

This section demonstrates how to reshape data using `stack()` and `unstack()` methods, which are powerful tools for working with hierarchical indices in Pandas DataFrames and Series.

```python
import numpy as np

data = np.arange(15).reshape((3,5))
indexers = ['Rainfall', 'Humidity', 'Wind']
dframe1 = pd.DataFrame(data, index=indexers, columns=['Bergen', 'Oslo', 'Trondheim', 'Stavanger', 'Kristiansand'])
print("Original DataFrame (dframe1):")
display(dframe1)
```

### Using `stack()`



```python
stacked = dframe1.stack()
print("Stacked Series:")
display(stacked)
```

### Using `unstack()`


```python
unstacked = stacked.unstack()
print("Unstacked DataFrame (back to original shape):")
display(unstacked)
```

### Concatenating Series with Hierarchical Keys and `unstack()`

```python
series1 = pd.Series([000, 111, 222, 333], index=['zeros','ones', 'twos', 'threes'])
series2 = pd.Series([444, 555, 666], index=['fours', 'fives', 'sixs'])

print("series1:")
display(series1)
print("\nseries2:")
display(series2)

frame2 = pd.concat([series1, series2], keys=['Number1', 'Number2'])
print("\nConcatenated Series with hierarchical keys:")
display(frame2)
```

```python
unstacked_frame2 = frame2.unstack()
print("Unstacked DataFrame from concatenated Series:")
display(unstacked_frame2)
```
